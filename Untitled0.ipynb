```python
#wczytanie pliku
import pandas as pd
df = pd.read_csv('bank_transactions.csv')
```

```python
#znalezienie uszkodzonych wierszy i zapamietanie id
uszkodzone_wiersze = df[['user_id', 'amount']].isna().any(axis=1)
uszkodzone_id = df[uszkodzone_wiersze]['user_id']
```

```python
#zapis do pliku id uszkodzonych wierszy
pd.DataFrame(uszkodzone_id, columns=['user_id']).to_csv('missing_report.csv', index=False)
```

```python
#usunięcie uszkodzonych wierszy
df_clean = df.dropna(subset=['amount', 'user_id']).copy()
```

```python
##### Podejrzane transakcje ######
```

```python
import ipaddress

zakresy_prywatne = [
    ipaddress.IPv4Network('10.0.0.0/8'),
    ipaddress.IPv4Network('172.16.0.0/12'),
    ipaddress.IPv4Network('192.168.0.0/16')
]

def czy_adres_prywatny(adres_ip):
  try:
    adres_ip = ipaddress.IPv4Address(adres_ip)
    for zakres in zakresy_prywatne:
      if adres_ip in zakres:
        return True
    return False
  except ipaddress.AddressValueError:
    return False
```

```python

podejrzane_wiersze = df_clean[(df_clean['amount'] > 15000) & df_clean['ip_address'].apply(czy_adres_prywatny)]
pd.DataFrame(podejrzane_wiersze).to_csv('fraud_alert.csv', index=False)
```

```python
#agregacja danych

#format na same daty
df_clean['timestamp'] = pd.to_datetime(df_clean['timestamp'].dt.date)
df_clean["month"] = df_clean["timestamp"].dt.to_period("M").astype(str)
```

```python
#grupowanie danych
df_grouped = df_clean.groupby(["category", "month"], as_index=False).agg({
    "amount": ["sum", "mean", "count"]
                                                                          })
df_grouped.to_csv("monthly_report.csv", index=False)
```
